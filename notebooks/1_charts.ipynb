```python
# === Define 1D → 2D manifold ===
def curve_1d_to_2d(t):
    return jnp.array([t[0], jnp.sin(t[0])])

man1 = manifolds.Manifold(1, 2, curve_1d_to_2d)

# === Define Charts as intervals in param space ===
boundary_U = (-jnp.pi + jnp.pi / 2, 0.0)
chart_U = charts.Chart(
    name="U",
    manifold=man1,
    boundary=boundary_U,
    chart_map=lambda t: t  # identity chart map
)

boundary_V = (-0.5, jnp.pi - jnp.pi / 2)
chart_V = charts.Chart(
    name="V",
    manifold=man1,
    boundary=boundary_V,
    chart_map=lambda t: -2*t
)

boundary_U_cap_V = charts.compute_chart_intersection(boundary_U, boundary_V)

chart_U_cap_V_X = charts.Chart(
    name="U_cap_V_X",
    manifold=man1,
    boundary=boundary_U_cap_V,
    chart_map=lambda t: t,
)

chart_U_cap_V_Y = charts.Chart(
    name="U_cap_V_Y",
    manifold=man1,
    boundary=boundary_U_cap_V,
    chart_map=lambda t: -2*t,
)

# === Plot manifold with chart overlays ===
fig, ax = plt.subplots(figsize=(10, 10))

plot_manifolds.plot_manifold(
    ax=ax, manifold=man1, resolution=100, xmin=-jnp.pi, xmax=jnp.pi, ymin=-jnp.pi, ymax=jnp.pi,
    color="skyblue", alpha=1.0, edgecolor="none", label=r"$\left(\mathcal{M}, \mathcal{O}, \mathcal{A}\right)$",
    label_position="center", label_fontsize=20, function=None, cmap="viridis", label_function="Scalar field $f$",
)

plot_charts.plot_chart_region_manifold(
    ax=ax, surface_manifold=man1, chart=chart_U,
    color="red", n_points=512, linestyle="dashed", linewidth=2.0,
    fill_surface=False, alpha=0.8,
    label=r"$\mathcal{U}$", label_position="center", label_fontsize=14,
    draw_endpoints=True, endpoint_color="red", endpoint_size=50,
)
plot_charts.plot_chart_region_manifold(
    ax=ax, surface_manifold=man1, chart=chart_V,
    color="blue", n_points=512, linestyle="dashed", linewidth=2.0,
    fill_surface=False, alpha=0.8,
    label=r"$\mathcal{V}$", label_position="center", label_fontsize=14,
    draw_endpoints=True, endpoint_color="blue", endpoint_size=50,
)
plot_charts.plot_chart_region_manifold(
    ax=ax, surface_manifold=man1, chart=chart_U_cap_V_X,
    color="purple", n_points=512, linestyle="dotted", linewidth=3.5,
    fill_surface=True, alpha=0.9, edgecolor=None, linewidth_fill=0.5,
    label=r"$\mathcal{U} \cap \mathcal{V}$", label_position="bottom", label_fontsize=14,
    draw_endpoints=True, endpoint_color="purple", endpoint_size=30,
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import jax.numpy as jnp
from jax import jacrev, vmap
from matplotlib.lines import Line2D
import numpy as np

# ------------------------------
# Define test points
# ------------------------------
param_points_U = jnp.array([[0.2, -0.5], [0.1, -0.6], [0.0, -0.7]])
param_points_V = jnp.array([
    [jnp.pi / 4, jnp.pi / 2],
    [jnp.pi / 6, jnp.pi / 3],
    [jnp.pi / 3, jnp.pi / 4],
])

# ------------------------------
# Compute chart images and Jacobians
# ------------------------------
chart_points_U = chart_U.map_to_chart(param_points_U)
chart_points_V = chart_V.map_to_chart(param_points_V)

jacobians_U = vmap(jacrev(chart_U.map_to_chart))(param_points_U)
jacobians_V = vmap(jacrev(chart_V.map_to_chart))(param_points_V)

# ------------------------------
# Plot chart Jacobian vectors
# ------------------------------
def plot_jacobian_vectors(points, jacobians, title):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.set_aspect('equal')
    ax.set_title(title)
    ax.set_xlabel("chart x")
    ax.set_ylabel("chart y")
    ax.grid(True)

    for p, J in zip(points, jacobians):
        x0, x1 = float(p[0]), float(p[1])
        J = np.array(J)
        ax.plot(x0, x1, 'ko')
        ax.quiver(x0, x1, J[0, 0], J[1, 0], color='red', scale=10)
        ax.quiver(x0, x1, J[0, 1], J[1, 1], color='blue', scale=10)

    legend_elements = [
        Line2D([0], [0], color='red', lw=2, label='Col 0 of Jacobian'),
        Line2D([0], [0], color='blue', lw=2, label='Col 1 of Jacobian'),
        Line2D([0], [0], marker='o', color='k', lw=0, label='Chart point')
    ]
    ax.legend(handles=legend_elements)
    plt.tight_layout()
    plt.show()

plot_jacobian_vectors(chart_points_U, jacobians_U, "Jacobian Vectors in Chart U")
plot_jacobian_vectors(chart_points_V, jacobians_V, "Jacobian Vectors in Chart V")

# ------------------------------
# Inverse map recovery
# ------------------------------
recovered_params_U = chart_U.inverse_map(chart_points_U)
recovered_params_V = chart_V.inverse_map(chart_points_V)

def plot_inverse_recovery(true_params, recovered_params, title):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.set_title(title)
    ax.set_xlabel(r"$u_1$")
    ax.set_ylabel(r"$u_2$")
    ax.set_aspect("equal")
    ax.grid(True)

    ax.plot(true_params[:, 0], true_params[:, 1], "ko", label="True param point")
    ax.plot(recovered_params[:, 0], recovered_params[:, 1], "gx", label="Recovered from inverse")

    for true, rec in zip(true_params, recovered_params):
        ax.plot([true[0], rec[0]], [true[1], rec[1]], "k--", alpha=0.4)

    ax.legend()
    plt.tight_layout()
    plt.show()

plot_inverse_recovery(param_points_U, recovered_params_U, "Inverse Recovery in param space — chart_U")
plot_inverse_recovery(param_points_V, recovered_params_V, "Inverse Recovery in param space — chart_V")

# ------------------------------
# Jacobian of inverse maps
# ------------------------------
def unbatched_inverse(chart_obj):
    def f(x):
        return chart_obj.inverse_map(x[None, :])[0]
    return f

jac_inv_U = vmap(jacrev(unbatched_inverse(chart_U)))(chart_points_U)
jac_inv_V = vmap(jacrev(unbatched_inverse(chart_V)))(chart_points_V)

def plot_inverse_jacobian(chart_points, jacobians, title):
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.set_title(title)
    ax.set_xlabel("chart x")
    ax.set_ylabel("chart y")
    ax.set_aspect("equal")
    ax.grid(True)

    for x, J in zip(chart_points, jacobians):
        x = np.array(x)
        J = np.array(J)
        if J.shape != (2, 2):
            continue  # or raise error

        x0, x1 = x[0], x[1]
        ax.plot(x0, x1, "ko")
        ax.quiver(x0, x1, J[0, 0], J[1, 0], color='red', scale=10)
        ax.quiver(x0, x1, J[0, 1], J[1, 1], color='blue', scale=10)

    legend_elements = [
        Line2D([0], [0], color='red', lw=2, label='Col 0 of inverse Jacobian'),
        Line2D([0], [0], color='blue', lw=2, label='Col 1 of inverse Jacobian'),
        Line2D([0], [0], marker='o', color='k', lw=0, label='Chart point')
    ]
    ax.legend(handles=legend_elements)
    plt.tight_layout()
    plt.show()

plot_inverse_jacobian(chart_points_U, jac_inv_U, "Jacobian of Inverse Map — chart_U")
plot_inverse_jacobian(chart_points_V, jac_inv_V, "Jacobian of Inverse Map — chart_V")
```
