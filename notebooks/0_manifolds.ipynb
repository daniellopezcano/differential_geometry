```python
import os
os.environ["JAX_PLATFORM_NAME"] = "cpu"

import numpy as np
import jax
import jax.numpy as jnp

import matplotlib
import matplotlib.pyplot as plt
from matplotlib import animation
from matplotlib.patches import FancyArrowPatch
import matplotlib.patheffects as path_effects
from matplotlib.lines import Line2D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# === Global Plotting Style ===
plt.style.use('tableau-colorblind10')
plt.rcParams.update({
    'text.usetex': True,
    'font.size': 16,
    'axes.linewidth': 1.2,
    'axes.edgecolor': 'black',
    'xtick.major.size': 5,
    'ytick.major.size': 5
})

import config.config as config

import differential_geometry.aux_tools as aux

import differential_geometry.manifolds as manifolds

import plotting_tools.plot_manifolds as plot_manifolds
```

```python
def curve_1d_to_2d(t):
    return jnp.array([t[0], jnp.sin(t[0])])
man1 = manifolds.Manifold(1, 2, curve_1d_to_2d)

fig, ax = plt.subplots(figsize=(10, 10))

plot_manifolds.plot_manifold(
    ax=ax, manifold=man1, resolution=100, xmin=-jnp.pi, xmax=jnp.pi, ymin=-jnp.pi, ymax=jnp.pi,
    color="skyblue", alpha=1.0, edgecolor="none", label=r"$\left(\mathcal{M}, \mathcal{O}, \mathcal{A}\right)$",
    label_position="center", label_fontsize=20, function=None, cmap="viridis", label_function="Scalar field $f$",
)

param_pts = jnp.array([[0.0], [jnp.pi / 2]])
embedded = man1.embed(param_pts)
jacobians = man1.derivatives_at_params(param_pts)
labels = [["$\\partial_0$"], ["$\\partial_1$"]]
plot_manifolds.plot_tangent_vectors(
    ax=ax, embedded_points=embedded, jacobians=jacobians, scale=0.5, colors=None, arrowstyle='-|>',
    mutation_scale=15, linewidth=1.5, alpha=1.0, draw_points=True, point_color="black", point_size=20,
    labels=labels, label_fontsize=20, label_offset=0.1
)
labels = [r"$T_p\mathcal{M}$", r"$T_q\mathcal{M}$"]
colors = ["purple", "darkorange"]
plot_manifolds.plot_tangent_spaces(
    ax=ax, embedded_points=embedded, jacobians=jacobians, size=1.6, colors=colors,
    alpha=0.5, resolution=10, line_width=2.0, labels=labels, label_fontsize=20,  label_offset=0.1
)

plt.tight_layout()
plt.show()



def curve_1d_to_3d(t):
    return jnp.array([jnp.cos(t[0]), jnp.sin(t[0]), t[0]])
man2 = manifolds.Manifold(1, 3, curve_1d_to_3d)

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

plot_manifolds.plot_manifold(
    ax=ax, manifold=man2, resolution=100, xmin=-jnp.pi, xmax=jnp.pi, ymin=-jnp.pi, ymax=jnp.pi,
    color="skyblue", alpha=1.0, edgecolor="none", label=r"$\left(\mathcal{M}, \mathcal{O}, \mathcal{A}\right)_d$",
    label_position="center", label_fontsize=20, function=None, cmap="viridis",  label_function="Scalar field $f$",
)

param_pts = jnp.array([[0.0], [jnp.pi / 2]])
embedded = man2.embed(param_pts)
jacobians = man2.derivatives_at_params(param_pts)
labels = [["$\\partial_1$"], ["$\\partial_2$"]]
plot_manifolds.plot_tangent_vectors(
    ax=ax, embedded_points=embedded, jacobians=jacobians, scale=0.5, colors=None, arrowstyle='-|>',
    mutation_scale=15, linewidth=1.5, alpha=1.0, draw_points=True, point_color="black", point_size=20,
    labels=labels, label_fontsize=20, label_offset=0.1
)
labels = [r"$T_p\mathcal{M}$", r"$T_q\mathcal{M}$"]
colors = ["purple", "darkorange"]
plot_manifolds.plot_tangent_spaces(
    ax=ax, embedded_points=embedded, jacobians=jacobians, size=1.6, colors=colors,
    alpha=0.5, resolution=10, line_width=2.0, labels=labels, label_fontsize=20, label_offset=0.1
)

ax.view_init(elev=50.0, azim=-60.0)
plt.tight_layout()
plt.show()



manifold = manifolds.Manifold(
    param_dim=2, ambient_dim=3,
    embedding_func=manifolds.factory_surface_embedding_gaussian_eggs(
        amplitude=1.0, center=(0.0, 0.0), sigma=(jnp.pi, jnp.pi)
    ),
)

xmin, xmax = jnp.array([-jnp.pi, jnp.pi]) * jnp.pi/3
ymin, ymax = jnp.array([-jnp.pi, jnp.pi]) * jnp.pi/3

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

plot_manifolds.plot_manifold(
    ax=ax, manifold=manifold, resolution=100, xmin=xmin, xmax=xmax, ymin=ymin, ymax=ymax,
    color="skyblue", alpha=0.2, edgecolor="none", label=r"$\left(\mathcal{M}, \mathcal{O}, \mathcal{A}\right)_d$",
    label_position="top", label_fontsize=20, function=None, cmap="viridis", label_function="Scalar field $f$",
)

param_pts = jnp.array([[0., 0.], [jnp.pi/2, jnp.pi/2]])
embedded = manifold.embed(param_pts)
jacobians = manifold.derivatives_at_params(param_pts)
labels = [
    ["$\\partial_1$", "$\\partial_2$"],
    ["$\\partial_1$", "$\\partial_2$"]
]
plot_manifolds.plot_tangent_vectors(
    ax=ax, embedded_points=embedded, jacobians=jacobians, scale=0.4, colors=None, arrowstyle='-|>',
    mutation_scale=15, linewidth=1.5, alpha=1.0, draw_points=True, point_color="black", point_size=20,
    labels=labels, label_fontsize=20, label_offset=0.1
)
labels = [r"$T_p\mathcal{M}$", r"$T_q\mathcal{M}$"]
colors = ["purple", "darkorange"]
plot_manifolds.plot_tangent_spaces(
    ax=ax, embedded_points=embedded, jacobians=jacobians, size=1.6, colors=colors,
    alpha=0.5, resolution=10, line_width=2.0, labels=labels, label_fontsize=20,  label_offset=0.4
)

ax.view_init(elev=50.0, azim=-60.0)
plt.tight_layout()
plt.show()
```
